```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Input
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
def polynomial(x):
    return 5*x**3 - 10*x**2 - 20*x + 10
```

```python
x = np.linspace(-20, 20, 5000).reshape(-1, 1)
y = polynomial(x)
```

```python
x_scaler = MinMaxScaler(feature_range=(-1, 1))
y_scaler = MinMaxScaler(feature_range=(-1, 1))

x_normalized = x_scaler.fit_transform(x)
y_normalized = y_scaler.fit_transform(y)
```

```python
plt.figure(figsize=(8, 5))
plt.plot(x, y)
plt.title("Original Polynomial")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(x_normalized, y_normalized)
plt.title("Normalized Polynomial")
plt.xlabel("Normalized x")
plt.ylabel("Normalized y")
plt.grid(True)
plt.show()
```

```python
x_train, x_temp, y_train, y_temp = train_test_split(
    x_normalized, y_normalized, test_size=0.20, random_state=42
)

x_val, x_test, y_val, y_test = train_test_split(
    x_temp, y_temp, test_size=0.50, random_state=42
)

print("Train:", len(x_train))
print("Validation:", len(x_val))
print("Test:", len(x_test))
```

```python
model = Sequential([
    Input(shape=(1,)),
    Dense(32, activation="relu"),
    Dense(64, activation="relu"),
    Dense(128, activation="relu"),
    Dense(1)
])

model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)
```

```python
history = model.fit(
    x_train,
    y_train,
    epochs=100,
    batch_size=32,
    validation_data=(x_val, y_val),
    verbose=1
)
```

```python
plt.figure(figsize=(8, 5))
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("Training vs Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(history.history["mae"], label="Training MAE")
plt.plot(history.history["val_mae"], label="Validation MAE")
plt.title("Training vs Validation MAE")
plt.xlabel("Epoch")
plt.ylabel("MAE")
plt.legend()
plt.grid(True)
plt.show()
```

```python
test_loss, test_mae = model.evaluate(x_test, y_test, verbose=0)

print("Test MSE:", test_loss)
print("Test MAE:", test_mae)
```

```python
y_pred = model.predict(x_test, verbose=0)
r2 = r2_score(y_test, y_pred)

print("R² Score:", r2)
```

```python
x_example = np.array([[5.0]])

x_example_normalized = x_scaler.transform(x_example)
y_pred_normalized = model.predict(x_example_normalized, verbose=0)
y_pred_actual = y_scaler.inverse_transform(y_pred_normalized)

actual_y = polynomial(x_example)

print("Input x:", x_example[0, 0])
print("Predicted y:", y_pred_actual[0, 0])
print("Actual y:", actual_y[0, 0])
print("Absolute Error:", abs(y_pred_actual[0, 0] - actual_y[0, 0]))
```
