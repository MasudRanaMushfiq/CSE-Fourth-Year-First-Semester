```python
import hashlib

def sha256(data):
    return hashlib.sha256(data.encode()).hexdigest()

def build_merkle_tree(leaves):
    if len(leaves) == 1:
        return leaves[0]

    if len(leaves) % 2 == 1:
        leaves.append(leaves[-1])

    new_level = []
    for i in range(0, len(leaves), 2):
        new_level.append(sha256(leaves[i] + leaves[i+1]))

    return build_merkle_tree(new_level)

# Input data (like transactions)
data_blocks = ["apple", "banana", "cherry", "date"]

# Hash the data blocks first
leaf_hashes = [sha256(data) for data in data_blocks]

# Get Merkle Root
merkle_root = build_merkle_tree(leaf_hashes)
print("Merkle Root:", merkle_root)
```
